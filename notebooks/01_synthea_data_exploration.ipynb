```python
from pathlib import Path

import pandas as pd

PROJECT_ROOT = Path.cwd().parent
CSV_DIR = PROJECT_ROOT / "data" / "raw" / "synthea_csv"

patients = pd.read_csv(CSV_DIR / "patients.csv", low_memory=False)
encounters = pd.read_csv(CSV_DIR / "encounters.csv", low_memory=False)
conditions = pd.read_csv(CSV_DIR / "conditions.csv", low_memory=False)
medications = pd.read_csv(CSV_DIR / "medications.csv", low_memory=False)
observations = pd.read_csv(CSV_DIR / "observations.csv", low_memory=False)

print("Patients:", len(patients))
print("Encounters:", len(encounters))
print("Conditions:", len(conditions))
print("Medications:", len(medications))
print("Observations:", len(observations))
```

```python
selected = pd.read_csv(PROJECT_ROOT / "data" / "selected_patients" / "selected_patient_cohort.csv")

selected.head(10)
```

```python
patient_id = selected.iloc[0]["patient_id"]

patient_encounters = encounters[encounters["PATIENT"] == patient_id].copy()

patient_medications = medications[medications["PATIENT"] == patient_id].copy()

patient_observations = observations[observations["PATIENT"] == patient_id].copy()

display(patient_encounters.head())
display(patient_medications.head())
display(patient_observations.head())
```
